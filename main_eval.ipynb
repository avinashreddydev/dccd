```python
import json
import yaml
from typing import Any


def pretty_print(data: Any, indent: int = 0, key_name: str | None = None):
    prefix = " " * indent

    # If this node has a key (from parent), print it
    if key_name is not None:
        if isinstance(data, (dict, list)):
            print(f"{prefix}{key_name}:")
        else:
            # Scalar value
            if isinstance(data, float):
                print(f"{prefix}{key_name}: {data:.2f}")
            else:
                print(f"{prefix}{key_name}: {data}")
            return

    # Dict
    if isinstance(data, dict):
        for k, v in data.items():
            pretty_print(v, indent + 2, k)

    # List
    elif isinstance(data, list):
        for i, item in enumerate(data):
            if isinstance(item, (dict, list)):
                print(f"{prefix}-")
                pretty_print(item, indent + 2)
            else:
                if isinstance(item, float):
                    print(f"{prefix}- {item:.2f}")
                else:
                    print(f"{prefix}- {item}")

    # Scalar (root-level)
    else:
        if isinstance(data, float):
            print(f"{prefix}{data:.2f}")
        else:
            print(f"{prefix}{data}")


def load_yaml_or_json(file_path: str):
    with open(file_path, "r") as f:
        if file_path.endswith((".yaml", ".yml")):
            return yaml.safe_load(f)
        else:
            return json.load(f)

```

```python
import os

results = []

for model in models:
    for alg in os.listdir(model):
        alg_path = os.path.join(model, alg)
        if not os.path.isdir(alg_path):
            continue

        for task in os.listdir(alg_path):
            task_path = os.path.join(alg_path, task)
            if task not in ["gsm_symbolic", "gsm8k", "math500", "prover9"]:
                continue
            file_path = os.path.join(task_path, "summary_results.json")

            if os.path.exists(file_path):
                data = load_yaml_or_json(file_path)


                print("=" * 80)
                print(f"Model: {model}")
                print(f"Algorithm: {alg}")
                print(f"Task: {task}")
                print("-" * 80)



                result = {
                    "model_name": model,
                    "algorithm": alg,
                    "task": task,
                    "data": data["accuracy"]
                }
                results.append(result)

                pretty_print(data)
```

```python


import pandas as pd
df = pd.DataFrame(results)
```

```python
table = df.pivot(
    index=["model_name", "algorithm"],   # rows
    columns="task",                      # new columns
    values="data"                        # cell values
)
```
